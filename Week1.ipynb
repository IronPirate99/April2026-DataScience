```python
# We will now compute the Mean, Median, and Mode for Age, G1, G2, G3. 
mean_age = df['age'].mean()
median_age = df['age'].median()
mode_age = df['age'].mode()
variance_age = df ['age'].var()
max_age = df ['age'].max()
min_age = df ['age'].min()
count_age = df['age'].count()

print("mean Age:", mean_age)
print("median Age:", median_age)
print("modal Age:", mode_age)
print("variance in Age:",variance_age)
print("max age", max_age)
print("min age", min_age)
print("count age", count_age)
```

```python
columns = ["age","G1","G2","G3"] 

for column in columns:
    print (f" Mean  {column} -   : {df[column].mean()}")
    print (f" Mode  {column} -   : {df[column].mode()}")
    print (f" Median  {column} - : {df[column].median()}")
    print (f" Variance {column} -: {df[column].var()}")
    print (f" Maximum {column} - : {df[column].max()}")
    print (f" Minimum {column} - : {df[column].min()}")
    print (f" Count {column} -   : {df[column].count()}")
    print("============================================")
```
